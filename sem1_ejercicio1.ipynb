```python
# Importar librerías
import duckdb
```

```python
# Crear la conexión a la db Chinook
con = duckdb.connect()
con.execute("INSTALL sqlite; LOAD sqlite;")
con.execute("ATTACH 'Chinook_Sqlite.sqlite' AS chinook (TYPE sqlite);")
```

```python
# Traemos la tabla "Invoices" mediante un dataframe
invoices = con.sql("""
    SELECT *
    FROM chinook.Invoice;
""").df()
invoices.head(5)
```

```python
# Visualizar las características de la tabla
invoices.describe()
```
