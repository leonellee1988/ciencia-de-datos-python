# **Comprehensions**

```python
# Datos:
generos = ["Rock", "Jazz", "Metal", "Pop", "Classical"]
precios = [0.99, 1.29, 0.99, 1.99, 0.99]
```

```python
# List comprehension: Construir una nueva lista
generos_cortos = [g.upper() for g in generos if len(g) <= 4]
generos_cortos
```

```python
# Dict comprehension: Pares clave-valor
precio_por_genero = {g: p for g, p in zip(generos, precios)}
precio_por_genero
```

```python
# Set comprehension: Valores únicos, sin duplicados
precios_unicos = {p for p in precios}
precios_unicos
```

```python
# Nota: repasando funciones lambda
cuadrado = lambda x: x**2
print(cuadrado(5))
```

# **Ejercicio 1.1**

Usando tu **conexión DuckDB** de la Semana 1, trae la tabla Track completa como lista de tuplas (con.sql("SELECT Name, Milliseconds, UnitPrice FROM chinook.Track").fetchall()) y, sin usar Pandas todavía, resuelve con comprehensions:

1. Una lista con los nombres de las pistas que duran más de 5 minutos (300,000 ms)

2. Un diccionario {nombre_pista: duracion_en_minutos}, redondeado a 1 decimal

3. El conjunto de precios únicos (UnitPrice) que existen en toda la tabla

```python
# Importar librerías
import duckdb
```

```python
# Realizar al conexión a la DB
con = duckdb.connect()
con.execute("INSTALL sqlite; LOAD sqlite;")
con.execute("ATTACH 'Chinook_Sqlite.sqlite' AS chinook (TYPE sqlite);")
```

```python
# Tabla track (como tuplas)
track_tb = con.sql("""
    SELECT Name, Milliseconds, UnitPrice
    FROM chinook.Track
""").fetchall()
print(len(track_tb))
```

```python
# Problema No. 1
tracks_over_5min = [t[0] for t in track_tb if t[1] > 300000]
print(len(tracks_over_5min))
tracks_over_5min
```

```python
# Problema No. 2
name_track = [n[0] for n in track_tb]
duration_track = [round(d[1] / 60000,1) for d in track_tb]
dict_track = {n: d for n, d in zip(name_track, duration_track)}
dict_track
```

```python
# Problema No. 3
price_track = [p[2] for p in track_tb]
unique_price = {p for p in price_track}
unique_price
```
